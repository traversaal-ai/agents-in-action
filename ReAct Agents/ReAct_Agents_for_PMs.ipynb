## 📚 Part 1: What is a ReAct Agent?

Before we write a single line of code, let's understand what we're building.

### The Old Way: A Simple LLM Call

Imagine you ask ChatGPT: *"What are the top 3 risks of launching our product in Germany next quarter?"*

ChatGPT gives you an answer based on its training data — but it **cannot**:
- Look up today's news
- Check your internal documents
- Run a calculation
- Ask a follow-up question and refine based on the answer

It's a **one-shot oracle** — brilliant, but passive.

### The New Way: A ReAct Agent

**ReAct** stands for **Re**asoning + **Act**ing. Published by researchers at Google in 2022, it describes a pattern where an AI:

1. **Thinks** about what it needs to do (Reasoning)
2. **Picks a tool** and uses it (Acting)
3. **Observes** the result
4. **Repeats** until the task is done

This loop is called the **Thought → Action → Observation** cycle.

```
┌─────────────────────────────────────────────────────────┐
│                    THE ReAct LOOP                       │
│                                                         │
│   GOAL: "Research competitor pricing in EU market"      │
│                          │                              │
│                          ▼                              │
│              ┌──────────────────────┐                   │
│              │   💭 THOUGHT         |                   │
│              │  "I need to search   │                   │
│              │   the web for this"  │                   │
│              └──────────┬───────────┘                   │
│                         │                               │
│                         ▼                               │
│              ┌──────────────────────┐                   │
│              │   ⚡ ACTION          │                   │
│              │  [calls web search   │                   │
│              │   tool]              │                   │
│              └──────────┬───────────┘                   │
│                         │                               │
│                         ▼                               │
│              ┌──────────────────────┐                   │
│              │   👁️ OBSERVATION     │                   │
│              │  "Found: Competitor  │                   │
│              │   X charges €49/mo"  │                   │
│              └──────────┬───────────┘                   │
│                         │                               │
│              ┌──────────▼───────────┐                   │
│              │  Done? No → Loop     │                   │
│              │  Done? Yes → Answer  │                   │
│              └──────────────────────┘                   │
└─────────────────────────────────────────────────────────┘
```

### Why Should PMs Care?

As a PM, you're constantly doing this loop yourself:
- *"I need to understand the competitive landscape"* → You Google, read, synthesize
- *"I need to know if the feature is feasible"* → You talk to engineering, take notes, refine
- *"I need to write the PRD"* → You pull together all the above

A ReAct agent **automates this research-and-synthesis workflow**. It's like having a tireless, fast research analyst on your team.

### Meet the Agents

| Agent | Role | What it does | Tools it uses |
|-------|------|-------------|---------------|
| 🔍 **Market Researcher** | Gathers raw data | Searches for market size, regulations, trends in target countries | Web search, text analysis |
| 🥊 **Competitive Analyst** | Profiles competitors | Identifies who's already in each market, their pricing, positioning | Web search, structured extraction |
| 📋 **Strategy Synthesizer** | Makes recommendations | Combines all research into a prioritized GTM recommendation | Summarization, scoring |

### Why Multi-Agent Instead of One Agent?

Think of it like a real team:
- A **generalist** agent doing everything gets confused and loses context
- **Specialized** agents each become expert in their domain
- Each agent has its own **memory** and **focus**
- They **check each other's work** through sequencing

This mirrors how you'd actually staff a project — you wouldn't ask your data analyst to also write the executive summary.

## 🔧 Part 4: Building the Blocks — Tools

Before we create our agents, we need to give them **tools** — capabilities they can call during their ReAct loops.

Think of tools like apps on your phone. The agent decides *when* to use them, just like you decide when to open Google Maps vs. calling a friend.

### The Tools Our Agents Will Use:

```
┌─────────────────────────────────────────────────────────┐
│                    AGENT TOOLBOX                        │
│                                                         │
│  🔍 SerperDevTool     → Google Search (real-time web)   │
│  🌐 ScrapeWebsiteTool → Read full webpage content       │
│  📊 Custom Tools      → Domain-specific calculations    │
│                                                         │
│  Agent picks the right tool for each step               │
│  like a surgeon choosing the right instrument           │
└─────────────────────────────────────────────────────────┘
```

> **💡 PM Insight:** When designing an agentic product, your key design decision is: *what tools should agents have access to?* More tools = more capability but more cost and risk. It's a tradeoff you'll manage as a PM.

### Creating Custom Tools

We'll build two custom tools for our scenario. In a real product, these would connect to live APIs (Statista, Crunchbase, Bloomberg, etc.).

For this class, we'll use **simulated data** — but the architecture and agent behavior is 100% real.

> **🎯 PM Learning:** Custom tools are where the real product differentiation lives. Anyone can give an agent a generic web search. Your competitive moat is proprietary data tools — connecting to your CRM, your analytics platform, your customer interview database.

````python
# ============================================================
# TOOL 1: Market Data Tool
# In production: connects to Statista, Eurostat, or your BI tool
# In this class: returns realistic simulated data
# ============================================================

class MarketDataTool(BaseTool):
    # Every tool needs a name and description
    # The agent reads these to decide WHEN to use this tool
    name: str = "Market Data Lookup"
    description: str = (
        "Retrieves market size, growth rate, regulatory environment, and "
        "digital adoption statistics for a given European country. "
        "Use this when you need quantitative market data for a specific country."
    )

    def _run(self, country: str) -> str:
        """
        This is what happens when the agent USES the tool.
        The agent passes in a country name, the tool returns data.
        """
        # Simulated market intelligence database
        # In production, this would be a real API call
        market_data = {
            "Germany": {
                "smb_count": "3.5 million",
                "expense_mgmt_market_size": "€420M (2024)",
                "yoy_growth": "18%",
                "digital_adoption_score": "72/100",
                "avg_time_to_regulatory_approval": "6-9 months",
                "key_regulation": "GDPR + GoBD (German accounting compliance)",
                "language_barrier": "High — German localization essential",
                "payment_preference": "SEPA Direct Debit dominant",
                "market_maturity": "Moderate — still 60% using spreadsheets",
                "risk_score": "Medium",
                "opportunity_score": "High"
            },
            "Netherlands": {
                "smb_count": "1.8 million",
                "expense_mgmt_market_size": "€180M (2024)",
                "yoy_growth": "24%",
                "digital_adoption_score": "89/100",
                "avg_time_to_regulatory_approval": "3-4 months",
                "key_regulation": "GDPR (well-established compliance culture)",
                "language_barrier": "Low — English widely accepted in business",
                "payment_preference": "iDEAL + SEPA",
                "market_maturity": "High — 75% open to SaaS tools",
                "risk_score": "Low",
                "opportunity_score": "Very High"
            },
            "France": {
                "smb_count": "4.1 million",
                "expense_mgmt_market_size": "€510M (2024)",
                "yoy_growth": "15%",
                "digital_adoption_score": "68/100",
                "avg_time_to_regulatory_approval": "8-12 months",
                "key_regulation": "GDPR + French e-invoicing mandate (2026)",
                "language_barrier": "Very High — French localization mandatory",
                "payment_preference": "CB (Carte Bancaire) + SEPA",
                "market_maturity": "Moderate — fragmented legacy solutions",
                "risk_score": "High",
                "opportunity_score": "High (long-term)"
            }
        }

        # Look up the requested country
        country_data = market_data.get(country)

        if country_data:
            # Format the data as readable text for the agent
            return f"Market Intelligence for {country}:\n" + "\n".join(
                [f"  • {key.replace('_', ' ').title()}: {value}"
                 for key, value in country_data.items()]
            )
        else:
            return f"No data available for {country}. Available markets: Germany, Netherlands, France"


print("✅ MarketDataTool created!")
print()

# Let's test it immediately so you can see what it returns
print("🧪 Testing the tool with 'Netherlands':")
print("-" * 50)
test_tool = MarketDataTool()
print(test_tool._run("Netherlands"))
````

## 🤝 Part 5: Creating the Agents

Now comes the exciting part — defining our agents! Each agent is essentially a **job description** for an AI employee:

- **Role**: Their job title
- **Goal**: What they're trying to achieve
- **Backstory**: Their "personality" and expertise context
- **Tools**: What capabilities they have access to
- **Verbose**: Whether we want to see their thinking process

The **backstory** might seem like creative writing, but it's actually critical — it sets the context that shapes how the LLM reasons. It's prompt engineering in disguise.

> **🎯 PM Insight:** Writing good agent backstories is a core product skill for agentic systems. It's like writing a job description AND an onboarding document simultaneously. Bad backstories = inconsistent agent behavior = unreliable product.

````python
# Initialize our tools (create instances we can hand to agents)
market_tool = MarketDataTool()
competitor_tool = CompetitorIntelligenceTool()

# ============================================================
# AGENT 1: THE MARKET RESEARCHER
# Responsible for: Gathering raw market intelligence
# ============================================================

market_researcher = Agent(
    role="Senior Market Research Analyst",

    goal=(
        "Gather comprehensive, data-driven market intelligence for Germany, "
        "Netherlands, and France. Focus on quantitative metrics: market size, "
        "growth rates, digital adoption, and regulatory complexity. "
        "Look up EACH country separately and compile a complete picture."
    ),

    backstory=(
        "You are a seasoned market research analyst with 12 years of experience "
        "in European B2B SaaS markets. You've worked at McKinsey and IDC Research "
        "before joining the strategy team at NovaPay. You are meticulous, data-driven, "
        "and always verify your findings. You know that a PM's go-to-market decision "
        "lives or dies on the quality of the underlying market data. "
        "You always look up data for EVERY country before drawing conclusions — "
        "never assume, always verify."
    ),

    tools=[market_tool],       # This agent can ONLY use the market data tool
    verbose=True,              # Show us its thinking (great for learning!)
    allow_delegation=False,    # This agent handles its own tasks
    max_iter=8                 # Maximum ReAct loop iterations
)

print("✅ Agent 1 created: Market Researcher")
print(f"   Role: {market_researcher.role}")
print(f"   Tools available: {[t.name for t in market_researcher.tools]}")
````

````python
# ============================================================
# AGENT 2: THE COMPETITIVE ANALYST
# Responsible for: Understanding who we'll compete against
# ============================================================

competitive_analyst = Agent(
    role="Competitive Intelligence Analyst",

    goal=(
        "Map the competitive landscape in Germany, Netherlands, and France "
        "for expense management software. Identify key competitors, their pricing, "
        "their weaknesses, and the gaps NovaPay could exploit. "
        "Be ruthlessly honest — do not sugarcoat competitive threats."
    ),

    backstory=(
        "You are a competitive intelligence specialist who previously worked at Gartner "
        "covering the fintech and expense management space. You have a talent for "
        "identifying the real reason customers switch products — not the marketing fluff, "
        "but the operational pain points. You believe that most startups fail in new "
        "markets not because the product is bad, but because they underestimated "
        "entrenched local competitors. Your job is to make sure NovaPay doesn't make "
        "that mistake. Always analyze ALL three countries before drawing conclusions."
    ),

    tools=[competitor_tool],   # This agent uses the competitor intelligence tool
    verbose=True,
    allow_delegation=False,
    max_iter=8
)

print("✅ Agent 2 created: Competitive Analyst")
print(f"   Role: {competitive_analyst.role}")
print(f"   Tools available: {[t.name for t in competitive_analyst.tools]}")
````

````python
# ============================================================
# AGENT 3: THE STRATEGY SYNTHESIZER
# Responsible for: Turning research into actionable recommendations
# ============================================================

strategy_synthesizer = Agent(
    role="Head of Product Strategy",

    goal=(
        "Synthesize market research and competitive intelligence into a clear, "
        "prioritized go-to-market recommendation for NovaPay's European expansion. "
        "Produce an executive-ready recommendation with a country priority ranking, "
        "entry strategy for each market, key risks, and a phased 18-month roadmap. "
        "Be specific and actionable — avoid vague consulting speak."
    ),

    backstory=(
        "You are a former VP of Product with experience scaling B2B fintech products "
        "across Europe. You've personally led market entries in 7 European countries "
        "and know the difference between a market that LOOKS attractive on paper "
        "versus one that actually generates revenue. You excel at translating dense "
        "research into clear strategic choices. You write for busy C-suite executives: "
        "top recommendation first, supporting evidence second, never bury the lead. "
        "You do NOT have access to tools — you synthesize from the research provided "
        "by your colleagues."
    ),

    tools=[],                  # No tools — this agent synthesizes, doesn't research
    verbose=True,
    allow_delegation=False,
    max_iter=5
)

print("✅ Agent 3 created: Strategy Synthesizer")
print(f"   Role: {strategy_synthesizer.role}")
print(f"   Tools available: None (synthesis only)")
````

## 📋 Part 6: Defining the Tasks

Agents are employees. **Tasks** are the actual work assignments.

A well-written task includes:
- **Description**: Detailed instructions (like a brief to a consultant)
- **Expected output**: Exactly what the deliverable should look like
- **Agent**: Who's responsible
- **Context** (optional): Which previous tasks this task depends on

> **🎯 PM Insight:** Task design is where most agentic systems fail. Vague tasks produce vague outputs. Think of this as writing an extremely detailed ticket in Jira — the agent is literal and will do exactly what you specify, no more.

````python
# ============================================================
# TASK 1: Market Research
# Assigned to: Market Researcher
# Depends on: Nothing (first in the chain)
# ============================================================

market_research_task = Task(
    description=(
        "Conduct a comprehensive market analysis for NovaPay's potential expansion "
        "into European expense management markets. \n\n"
        "REQUIRED STEPS:\n"
        "1. Use the Market Data Lookup tool to retrieve data for Germany\n"
        "2. Use the Market Data Lookup tool to retrieve data for Netherlands\n"
        "3. Use the Market Data Lookup tool to retrieve data for France\n"
        "4. Analyze and compare the three markets across: market size, growth rate, "
        "   digital adoption, regulatory complexity, and language requirements\n"
        "5. Rank the markets by overall opportunity score with clear reasoning\n\n"
        "Do NOT skip any country. The CEO needs data on all three."
    ),

    expected_output=(
        "A structured market research report containing:\n"
        "- Individual country profiles (Germany, Netherlands, France) with all key metrics\n"
        "- A comparison table highlighting key differences\n"
        "- A preliminary market opportunity ranking (1st, 2nd, 3rd) with written rationale\n"
        "- Specific risks or flags for each market\n"
        "Format: Clear sections with headers, bullet points for data, prose for analysis."
    ),

    agent=market_researcher     # Assigned to our first agent
)

print("✅ Task 1 created: Market Research")
````

````python
# ============================================================
# TASK 2: Competitive Analysis
# Assigned to: Competitive Analyst
# Depends on: Task 1 (needs to know which markets to analyze)
# ============================================================

competitive_analysis_task = Task(
    description=(
        "Based on the market research already conducted, perform a deep competitive "
        "intelligence analysis for Germany, Netherlands, and France.\n\n"
        "REQUIRED STEPS:\n"
        "1. Use the Competitor Intelligence tool for Germany\n"
        "2. Use the Competitor Intelligence tool for Netherlands\n"
        "3. Use the Competitor Intelligence tool for France\n"
        "4. For each country, identify: top 3 competitors, their pricing, key weaknesses,"
        "   and the specific gap NovaPay could fill\n"
        "5. Assess competitive intensity: is this a winner-take-all market or "
        "   fragmented enough for a new entrant?\n"
        "6. Identify which market has the LOWEST competitive barrier to entry for NovaPay\n\n"
        "Be specific about pricing. Executives need numbers, not vague descriptions."
    ),

    expected_output=(
        "A competitive intelligence report containing:\n"
        "- Per-country competitor profiles (top 3 per market) with pricing and weaknesses\n"
        "- A 'competitive moat' assessment for each country (easy/medium/hard to enter)\n"
        "- Identified whitespace opportunities (what's missing in each market)\n"
        "- NovaPay's potential differentiation angle in each country\n"
        "- A ranking of markets by ease of competitive entry (easiest first)\n"
        "Format: Clear sections, competitor comparison tables where helpful."
    ),

    agent=competitive_analyst,
    context=[market_research_task]  # This task READS the output of Task 1
)

print("✅ Task 2 created: Competitive Analysis")
print("   ℹ️  This task will receive Task 1's output as context")
````

````python
# ============================================================
# TASK 3: Strategic Synthesis
# Assigned to: Strategy Synthesizer
# Depends on: Both Task 1 AND Task 2
# ============================================================

strategy_synthesis_task = Task(
    description=(
        "You have received market research (Task 1) and competitive analysis (Task 2) "
        "for NovaPay's European expansion into Germany, Netherlands, and France.\n\n"
        "Your job is to synthesize these findings into a definitive go-to-market "
        "recommendation that the CEO can present to the board.\n\n"
        "REQUIRED OUTPUTS:\n"
        "1. Executive Summary (3-4 sentences): Bottom-line recommendation\n"
        "2. Country Priority Ranking: Which to enter first, second, third and WHY\n"
        "3. Entry Strategy per Country: Specific tactics (e.g., partner with local "
        "   accounting software, focus on SMBs under 50 employees, etc.)\n"
        "4. Risk Assessment: Top 2 risks per country and mitigation approach\n"
        "5. 18-Month Phased Roadmap: Quarter-by-quarter milestones\n"
        "6. Success Metrics: What KPIs to track to know if we're winning\n\n"
        "The CEO's time is precious. Start with the recommendation, then the evidence."
    ),

    expected_output=(
        "A complete Go-To-Market Strategy Document for NovaPay's European Expansion, "
        "formatted as a professional strategic brief with:\n"
        "- Executive Summary (the 'so what')\n"
        "- Country Priority Ranking (1-2-3) with rationale\n"
        "- Entry strategy per country (specific, actionable)\n"
        "- Risk register with mitigations\n"
        "- 18-month phased roadmap by quarter\n"
        "- KPIs and success metrics\n"
        "Tone: Confident, direct, executive-ready. No vague consulting speak."
    ),

    agent=strategy_synthesizer,
    context=[market_research_task, competitive_analysis_task]  # Reads BOTH prior tasks
)

print("✅ Task 3 created: Strategic Synthesis")
print("   ℹ️  This task receives outputs from BOTH Task 1 and Task 2")
````

## 🚀 Part 7: Assembling & Running the Crew

Now we assemble everything into a **Crew** — the team that will actually execute.

### Understanding Process Types

CrewAI supports different orchestration strategies:

```
SEQUENTIAL (what we're using):
Task 1 → Task 2 → Task 3
Each task waits for the previous one ✓
Good for: dependent research pipelines

HIERARCHICAL (advanced):
              Manager Agent
             /      |      \
        Task 1   Task 2   Task 3
A 'manager' agent delegates and reviews
Good for: complex decision-making workflows
```

> **⚠️ Cost Warning:** Running agents with real LLMs costs money. This notebook will use approximately **$0.15-0.40 in OpenAI API credits**.

````python
# ============================================================
# ASSEMBLE THE CREW
# ============================================================

novapay_intelligence_crew = Crew(
    # The team
    agents=[market_researcher, competitive_analyst, strategy_synthesizer],

    # The work to be done
    tasks=[market_research_task, competitive_analysis_task, strategy_synthesis_task],

    # Sequential: complete each task in order before moving to next
    process=Process.sequential,

    # Show detailed logs of the crew's work
    verbose=True
)

print("✅ Crew assembled!")
print()
print("Team lineup:")
print("  1. Market Researcher    → Task: Market Analysis (Germany, NL, France)")
print("  2. Competitive Analyst  → Task: Competitive Intelligence")
print("  3. Strategy Synthesizer → Task: GTM Recommendation")
print()
print("Workflow: Sequential (1 → 2 → 3)")
print()
print("🚀 Ready to launch! Run the next cell to start...")
````

````python
# ============================================================
# LAUNCH THE CREW
# ============================================================
# This is the moment everything comes together!
# Watch the verbose output — you'll see the ReAct loop in action:
#   THOUGHT: "I need to look up Germany first..."
#   ACTION: [calls Market Data Lookup tool with "Germany"]
#   OBSERVATION: [receives the data]
#   THOUGHT: "Now I need Netherlands..."
#   ... and so on

print("🚀 LAUNCHING NOVAPAY INTELLIGENCE CREW...")
print("=" * 60)
print("Watch the agents think, act, and observe in real-time below:")
print("=" * 60)
print()

# This is the single line that runs the entire agentic system!
result = novapay_intelligence_crew.kickoff()
````

## 🔬 Part 9: Understanding What Just Happened

Let's break down the ReAct loop you just witnessed:

### The Thinking Process (What the Agent Did)

```
AGENT 1 — Market Researcher ReAct Loop:
─────────────────────────────────────────────────────────
THOUGHT 1: "I need market data for Germany first"
ACTION 1:  market_data_tool("Germany")
OBSERVE 1: [receives Germany data]

THOUGHT 2: "Now I need Netherlands data"
ACTION 2:  market_data_tool("Netherlands")
OBSERVE 2: [receives Netherlands data]

THOUGHT 3: "Now France"
ACTION 3:  market_data_tool("France")
OBSERVE 3: [receives France data]

THOUGHT 4: "I have all the data. Time to analyze and write the report."
ACTION 4:  [writes comprehensive analysis — no tool needed]
FINAL:     [structured market research report]
─────────────────────────────────────────────────────────

AGENT 2 — Competitive Analyst ReAct Loop:
─────────────────────────────────────────────────────────
[Receives Agent 1's report as context]
THOUGHT 1: "I need competitor data for all three markets"
ACTION 1:  competitor_tool("Germany")
OBSERVE 1: [receives Germany competitor landscape]
... [repeats for NL and France]
FINAL:     [competitive intelligence report]
─────────────────────────────────────────────────────────

AGENT 3 — Strategy Synthesizer:
─────────────────────────────────────────────────────────
[Receives BOTH Agent 1 and Agent 2 reports]
THOUGHT 1: "Let me synthesize market opportunity + competitive ease"
[No tools needed — pure reasoning and synthesis]
FINAL:     [GTM Strategy Document]
─────────────────────────────────────────────────────────
```

### Key Observations for PMs

Notice how the agents:
1. **Planned before acting** — they didn't randomly call tools
2. **Used context** — each agent built on previous agents' work
3. **Self-corrected** — if a tool returned unexpected data, they adapted
4. **Stayed in their lane** — the Strategy Synthesizer didn't try to do research

### Experiment 2: Add a New Tool

Let's add a **Pricing Strategy Calculator** tool — something an agent can use to recommend optimal pricing for each market.

````python
# ============================================================
# EXPERIMENT: Build Your Own Tool
# This shows how extensible the system is
# ============================================================

class PricingStrategyTool(BaseTool):
    name: str = "Pricing Strategy Calculator"
    description: str = (
        "Calculates recommended pricing strategy for NovaPay in a given market. "
        "Inputs the country name and returns a pricing recommendation including "
        "suggested price point, positioning strategy (premium/value/penetration), "
        "and expected margin at scale. Use when you need pricing guidance for a market."
    )

    def _run(self, country: str) -> str:
        pricing_strategies = {
            "Netherlands": {
                "recommended_price": "€11/user/month",
                "strategy": "Penetration pricing — enter below Payhawk at €9, "
                            "add iDEAL integration as free differentiator, "
                            "raise to €13 after 6 months with premium tier",
                "competitive_position": "Price parity with market leader, superior features",
                "gross_margin_at_scale": "~74%",
                "payback_period": "8 months (industry avg: 14)",
                "freemium_viable": "Yes — high digital adoption supports self-serve"
            },
            "Germany": {
                "recommended_price": "€13/user/month",
                "strategy": "Value positioning — price between Moss (€10) and Spendesk (€15). "
                            "Lead with GoBD compliance as the hook — this removes a €15K/year "
                            "compliance consultant cost for German SMBs.",
                "competitive_position": "Mid-market, compliance-first messaging",
                "gross_margin_at_scale": "~71%",
                "payback_period": "12 months",
                "freemium_viable": "No — German enterprises expect full demos and trials"
            },
            "France": {
                "recommended_price": "€12/user/month",
                "strategy": "Premium positioning — cannot undercut Spendesk (signals inferiority). "
                            "Must justify price through e-invoicing mandate compliance (2026) "
                            "and multi-country capabilities Spendesk lacks.",
                "competitive_position": "Premium alternative for growth-stage companies",
                "gross_margin_at_scale": "~69%",
                "payback_period": "18 months",
                "freemium_viable": "No — relationship-driven sales culture"
            }
        }

        data = pricing_strategies.get(country)
        if data:
            return f"Pricing Strategy for {country}:\n" + "\n".join(
                [f"  • {k.replace('_', ' ').title()}: {v}" for k, v in data.items()]
            )
        return f"No pricing data available for {country}"


# Test the new tool
pricing_tool = PricingStrategyTool()
print("✅ New tool created: PricingStrategyTool")
print()
print("🧪 Test output for Netherlands:")
print("-" * 50)
print(pricing_tool._run("Netherlands"))
print()
print("💡 Now you could give this tool to the Strategy Synthesizer agent")
print("   so it can look up pricing recommendations during synthesis!")
print()
print("Try it: Add 'pricing_tool' to strategy_synthesizer's tools list")
print("and rebuild the crew. How does the output change?")
````

## 🧠 Part 11: PM Design Principles for Agentic Systems

Let's zoom out and extract the lessons you should take back to your product work.

````python
# A summary of key PM lessons — print it as a reference card

lessons = """
╔══════════════════════════════════════════════════════════════════════════╗
║         PM DESIGN PRINCIPLES FOR AGENTIC AI SYSTEMS                      ║
║         (From: Agentic AI System Design — Module 6)                      ║
╠══════════════════════════════════════════════════════════════════════════╣
║                                                                          ║
║  1. SPECIALIZE YOUR AGENTS                                               ║
║     One agent per domain > one agent for everything                      ║
║     Like hiring: you want experts, not generalists for complex work      ║
║                                                                          ║
║  2. DESIGN TASK DESCRIPTIONS LIKE BRILLIANT BRIEFS                       ║
║     Vague task → vague output. Every word in your task description       ║
║     matters. This is your highest-leverage design decision.              ║
║                                                                          ║
║  3. TOOLS ARE YOUR COMPETITIVE MOAT                                      ║
║     Generic agents + proprietary tools = differentiated product.         ║
║     What unique data sources can you connect to?                         ║
║                                                                          ║
║  4. THE GOAL DEFINES EVERYTHING                                          ║
║     Same data + different agent goal = totally different output.         ║
║     Define the RIGHT goal for the user's actual need, not the surface.   ║
║                                                                          ║
║  5. VERBOSE FIRST, OPTIMIZE LATER                                        ║
║     Always build with verbose=True to understand agent behavior.         ║
║     Turn off verbosity only once you've validated the system works.      ║
║                                                                          ║
║  6. COST IS A PRODUCT METRIC                                             ║
║     Every LLM call costs money. Monitor token usage. Set max_iter.       ║
║     Agent cost = COGS for agentic products. Optimize like infrastructure.║
║                                                                          ║
║  7. AGENTS NEED GUARDRAILS, NOT JUST GOALS                               ║
║     Add constraints (max_iter, specific output formats) to keep          ║
║     agents focused. Unconstrained agents go down rabbit holes.           ║
║                                                                          ║
║  8. HUMAN-IN-THE-LOOP IS A FEATURE, NOT A FAILURE                        ║
║     The best agentic systems know when to pause and ask a human.         ║
║     Design checkpoints for high-stakes decisions.                        ║
║                                                                          ║
║  9. BACKSTORY IS PROMPT ENGINEERING IN DISGUISE                          ║
║     The agent's backstory shapes its 'personality' and reasoning style.  ║
║     Invest time writing it — it's the closest thing to training data.    ║
║                                                                          ║
║  10. THE ReAct LOOP IS OBSERVABLE (AND AUDITABLE)                        ║
║      Unlike a black-box model, you can watch every Thought-Action-Obs.   ║
║      This is a huge advantage for trust, debugging, and compliance.      ║
║                                                                          ║
╚══════════════════════════════════════════════════════════════════════════╝
"""

print(lessons)
````
